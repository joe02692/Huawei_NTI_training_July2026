```python
import pandas as P
import seaborn as S
import numpy as N
import matplotlib.pyplot as MAT
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
Data_titanic = P.read_csv('/content/titanic_data (1).csv')
```

```python
display(Data_titanic)
```

```python
Data_titanic.isnull().sum()
```

```python
age_str = SimpleImputer(strategy='mean')
Data_titanic['age'] = age_str.fit_transform(Data_titanic[['age']])
```

```python
deck_str = SimpleImputer(strategy='most_frequent')
Data_titanic['deck'] = deck_str.fit_transform(Data_titanic[['deck']]).ravel()
```

```python
M.figure(figsize=(12, 6))
numeric_cols = Data_titanic.select_dtypes(include=[N.number]).columns
S.boxplot(data=Data_titanic[numeric_cols], palette='Set3')
M.title('Boxplot of Titanic Dataset Numeric Features', fontsize=15)
M.xticks(rotation=45)
M.grid(axis='y', linestyle='--', alpha=0.7)
M.show()
```

```python
for col in ['fare']:
    Q1 = Data_titanic[col].quantile(0.25)
    Q3 = Data_titanic[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    NO_outliers = Data_titanic[(Data_titanic[col] < lower_bound) | (Data_titanic[col] > upper_bound)]
```

```python
M.figure(figsize=(12, 6))
numeric_cols = NO_outliers.select_dtypes(include=[N.number]).columns
S.boxplot(data=NO_outliers[numeric_cols], palette='Set3')
M.title('Boxplot of Titanic Dataset Numeric Features', fontsize=15)
M.xticks(rotation=45)
M.grid(axis='y', linestyle='--', alpha=0.7)
M.show()
```

```python
LE = LabelEncoder()

for i in non_numeric :
  Data_titanic[i] = LE.fit_transform(Data_titanic[i])
```

```python
plot_df = Data_titanic.copy()
plot_df['survived_label'] = plot_df['survived'].map({0: 'Died', 1: 'Survived'})
plot_df['sex_label'] = plot_df['sex'].map({1: 'Male', 0: 'Female'})  # Based on standard LabelEncoding for this dataset
plot_df['class_label'] = plot_df['pclass'].map({1: '1st Class', 2: '2nd Class', 3: '3rd Class'})

MAT.figure(figsize=(14, 5))

# Plot 1: Survival Count
MAT.subplot(1, 3, 1)
S.countplot(x='survived_label', data=plot_df, palette='viridis', hue='survived_label', legend=False)
MAT.title('Total Survival Count')
MAT.xlabel('Status')

# Plot 2: Survival Rate by Sex
MAT.subplot(1, 3, 2)
S.barplot(x='sex_label', y='survived', data=plot_df, palette='magma', hue='sex_label', legend=False)
MAT.title('Survival Rate by Sex')
MAT.xlabel('Sex')
MAT.ylabel('Survival Probability')

# Plot 3: Survival by Class
MAT.subplot(1, 3, 3)
S.barplot(x='class_label', y='survived', data=plot_df, palette='plasma', hue='class_label', legend=False)
MAT.title('Survival Rate by Pclass')
MAT.xlabel('Passenger Class')
MAT.ylabel('Survival Probability')

MAT.tight_layout()
MAT.show()
```
